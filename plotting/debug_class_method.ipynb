```python
import numpy as np
import matplotlib.pyplot as plt
from slabguide import Slab, Propagator, Plot3D, real_contour

%matplotlib widget
plt.style.use('dark_background')
```

```python
A = Slab(Ts=[6, 3, 6])
```

```python
half_width = .8
shift = .75
amplitude = 1

Lx = -half_width + shift
Rx = half_width + shift

def quadratic_piecewise(x):
    x = np.array(x, dtype=float)
    conds = [(x < Lx), ((Lx <= x)*(x <= Rx)), (x > Rx)]
    funcs = [lambda x: 0, lambda x: amplitude * (1 - ((x - shift) / half_width)**2), lambda x: 0]
    return np.piecewise(x, conds, funcs)

f0 = quadratic_piecewise
```

```python
# A.plot_refractive_index();
# A.add_1d_plot(f0);
# plt.axis('equal');
```

```python
P = Propagator(A, f0=f0, window=[Lx,Rx])
```

```python
P.add_contour(real_contour(.00001, A.Z_evanescent, 150))
P.add_contour(real_contour(A.Z_evanescent, 4 * A.Z_evanescent, 150))
```

## Multiple Frequencies

```python
B = Slab(Ts=[6, 3, 6], wl=1.3e-6)
C = Slab(Ts=[6, 3, 6], wl=1e-6)
```

```python
Q = Propagator(B, f0=f0, window=[Lx,Rx])
R = Propagator(C, f0=f0, window=[Lx,Rx])
```

```python
Q.add_contour(real_contour(.00001, B.Z_evanescent, 150))
Q.add_contour(real_contour(B.Z_evanescent, 4 * B.Z_evanescent, 150))
```

```python
R.add_contour(real_contour(.00001, C.Z_evanescent, 150))
R.add_contour(real_contour(C.Z_evanescent, 4 * C.Z_evanescent, 150))
```

```python
Plot3D(P, Q, R, time_check=False)
```
