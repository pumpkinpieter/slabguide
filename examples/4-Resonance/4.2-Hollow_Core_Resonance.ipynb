# Propagation of Hollow Core waveguide at resonant and non-resonant wavelengths

Here we compare the fields induced in the hollow core guide at resonant and non-resonant operating wavelengths.

```python
import numpy as np
import matplotlib.pyplot as plt
from slabguide import Slab, Propagator, real_contour, Plot3D

plt.style.use('dark_background')
%matplotlib widget
```

# Setup Waveguides and input field

```python
A = Slab(scale=1e-5, Ts=[6.5, 1, 6, 1, 6.5],  ns=[1, 1.44, 1, 1.44, 1],
                 xrefs=[100, 70, 300, 70, 100], wl=5.83e-06)
```

```python
M = 10
res_wls = np.array([2 * A.n0 * A.ts[1] / m * np.sqrt((A.n_high/A.n0)**2 - 1) for m in range(1, M)])
```

```python
res_wls
```

```python
B = Slab(scale=1e-5, Ts=[6.5, 1, 6, 1, 6.5],  ns=[1, 1.44, 1, 1.44, 1],
                 xrefs=[100, 70, 300, 70, 100], wl=6.90764471839393e-06)
```

```python
half_width = 1
shift = .5
amplitude = 1

Lx = -half_width + shift
Rx = half_width + shift

def quadratic_piecewise(x):
    x = np.array(x, dtype=float)
    conds = [(x < Lx), ((Lx <= x)*(x <= Rx)), (x > Rx)]
    funcs = [lambda x: 0, lambda x: amplitude * (1 - ((x - shift) / half_width)**2), lambda x: 0]
    return np.piecewise(x, conds, funcs)

f0 = quadratic_piecewise
```

# Propagation

## Non-Resonant

```python
P = Propagator(A, f0=f0, window=[Lx, Rx], calculate_guided_field=False)
```

```python
Np = 100
Lz = .001
Rz = 1
P.add_contour(real_contour(Lz, Rz, Np))
```

```python
# P.plot_transform();
```

```python
Np = 500
Lz = 1
Rz = A.Z_evanescent

P.add_contour(real_contour(Lz, Rz, Np))
```

```python
P.add_contour(real_contour(.515, .525, 80))
P.add_contour(real_contour(1.02, 1.06, 40))
P.add_contour(real_contour(1.53, 1.58, 40))
P.add_contour(real_contour(2.05, 2.1, 40))
```

```python
Np = 100
Lz = A.Z_evanescent
Rz = 30

P.add_contour(real_contour(Lz, Rz, Np))
```

## Resonant

```python
Q = Propagator(B, f0=f0, window=[Lx, Rx], calculate_guided_field=False)
```

```python
Lz = .001
Rz = B.Z_evanescent
Np = 800

Q.add_contour(real_contour(Lz, Rz, Np))
```

```python
Lz = B.Z_evanescent
Rz = 30
Np = 200

Q.add_contour(real_contour(Lz, Rz, Np))
```
